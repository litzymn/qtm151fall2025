```python
import pandas as pd
import matplotlib.pyplot as plt

df1 = pd.read_csv("Random_Data1.csv")
df2 = pd.read_csv("Ordered_Data.csv")



df1_long = pd.melt(df1,
                var_name = "Sorting Method",
                value_name = "Time",
                id_vars = "N",
                value_vars = [" BubbleSort", "BubbleSortOpt",
                              "SelectionSort","InsertionSort","QuickSort",
                              "QuickSortOpt","MergeSortRec","MergeSortNon"])

df2_long = pd.melt(df2,
                var_name = "Sorting Method",
                value_name = "Time",
                id_vars = "N",
                value_vars = [" BubbleSort", "BubbleSortOpt",
                              "SelectionSort","InsertionSort","QuickSort",
                              "QuickSortOpt","MergeSortRec","MergeSortNon"])



fig, ax = plt.subplots(figsize=(12,7))

ax.plot(df1_long[df1_long["Sorting Method"]=="BubbleSort"]["N"],
        df1_long[df1_long["Sorting Method"]=="BubbleSort"]["Time"], 
        color="r", label="BubbleSort")

ax.plot(df1_long[df1_long["Sorting Method"]=="BubbleSortOpt"]["N"],
        df1_long[df1_long["Sorting Method"]=="BubbleSortOpt"]["Time"], 
        color="r", linestyle="--", label="BubbleSortOpt")

ax.plot(df1_long[df1_long["Sorting Method"]=="SelectionSort"]["N"],
        df1_long[df1_long["Sorting Method"]=="SelectionSort"]["Time"], 
        color="b", label="SelectionSort")

ax.plot(df1_long[df1_long["Sorting Method"]=="InsertionSort"]["N"],
        df1_long[df1_long["Sorting Method"]=="InsertionSort"]["Time"], 
        color="m", label="InsertionSort")

ax.plot(df1_long[df1_long["Sorting Method"]=="QuickSort"]["N"],
        df1_long[df1_long["Sorting Method"]=="QuickSort"]["Time"], 
        color="g", label="QuickSort")

ax.plot(df1_long[df1_long["Sorting Method"]=="QuickSortOpt"]["N"],
        df1_long[df1_long["Sorting Method"]=="QuickSortOpt"]["Time"], 
        color="c", linestyle="--", label="QuickSortOpt")

ax.plot(df1_long[df1_long["Sorting Method"]=="MergeSortRec"]["N"],
        df1_long[df1_long["Sorting Method"]=="MergeSortRec"]["Time"], 
        color="y", label="MergeSortRec")

ax.plot(df1_long[df1_long["Sorting Method"]=="MergeSortNon"]["N"],
        df1_long[df1_long["Sorting Method"]=="MergeSortNon"]["Time"], 
        color="k", linestyle="--", label="MergeSortNon")

ax.set_yscale("log")
ax.set_xlabel("N")
ax.set_ylabel("Time (seconds)")
ax.set_title("Sorting Algorithm Times — Random (Average Case)")
ax.grid(True)
ax.legend()
plt.show()

```
